```python

fig = plt.figure(figsize=(16, 12))

# Create dashboard layout
gs = fig.add_gridspec(3, 4)

# ---------------- KPI CARDS ----------------

ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[0, 2])
ax4 = fig.add_subplot(gs[0, 3])

# Total Sales
ax1.text(0.5, 0.5, f"₦{total_sales:,.2f}",
         fontsize=16, fontweight="bold", ha="center")
ax1.set_title("TOTAL SALES")
ax1.axis("off")

# Total Profit
ax2.text(0.5, 0.5, f"₦{total_profit:,.2f}",
         fontsize=16, fontweight="bold", ha="center")
ax2.set_title("TOTAL PROFIT")
ax2.axis("off")

# Average Order Value
ax3.text(0.5, 0.5, f"₦{aov:,.2f}",
         fontsize=16, fontweight="bold", ha="center")
ax3.set_title("AVERAGE ORDER VALUE")
ax3.axis("off")

# Top Product
ax4.text(0.5, 0.5, top_product,
         fontsize=16, fontweight="bold", ha="center")
ax4.set_title("TOP-SELLING PRODUCT")
ax4.axis("off")


# ---------------- CHARTS ----------------

# Sales by Product
ax5 = fig.add_subplot(gs[1, 0:2])
product_sales.plot(kind="bar", ax=ax5)
ax5.set_title("Sales by Product")
ax5.tick_params(axis="x", rotation=45)

# Sales by City
ax6 = fig.add_subplot(gs[1, 2:4])
city_sales.plot(kind="bar", ax=ax6)
ax6.set_title("Sales by City")
ax6.tick_params(axis="x", rotation=45)

# Sales by Category
ax7 = fig.add_subplot(gs[2, 0:2])
category_sales.plot(kind="bar", ax=ax7)
ax7.set_title("Sales by Category")
ax7.tick_params(axis="x", rotation=45)

# Monthly Sales Trend
ax8 = fig.add_subplot(gs[2, 2:4])
monthly_sales.plot(kind="line", marker="o", ax=ax8)
ax8.set_title("Monthly Sales Trend")
ax8.tick_params(axis="x", rotation=45)
fig.suptitle(
"SALES PERFORMANCE DASHBOARD\n2025",
fontsize=22,
fontweight="bold"
)
plt.tight_layout(rect=[0,0,1,0.95])
plt.savefig("sales_dashboard.png", dpi=300, bbox_inches="tight")
plt.show()
```
